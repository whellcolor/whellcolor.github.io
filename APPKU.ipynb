```python
from kivy.app import App
from kivy.uix.boxlayout import BoxLayout
from kivy.uix.label import Label
from kivy.uix.button import Button
from kivy.graphics import Color, Rectangle

class AppKuApp(App):
    def build(self):
        # Warna latar belakang aplikasi
        self.title = "APPKU"

        root = BoxLayout(orientation='vertical', padding=30, spacing=20)

        # Background putih bersih
        with root.canvas.before:
            Color(0.95, 0.95, 0.95, 1)
            self.rect = Rectangle(size=root.size, pos=root.pos)
        root.bind(size=self._update_rect, pos=self._update_rect)

        # Judul Aplikasi
        title_label = Label(
            text="[b]Selamat Datang di APPKU![/b]",
            markup=True,
            font_size='24sp',
            color=(0.1, 0.1, 0.1, 1),
            size_hint_y=None,
            height=50
        )
        root.add_widget(title_label)

        # Info Status AdMob & Paket
        info_text = (
            "Package: com.sd.app\n"
            "AdMob Status: Terhubung\n"
            "App ID: ca-app-pub-7182145371991939~2794760314"
        )
        info_label = Label(
            text=info_text,
            font_size='16sp',
            color=(0.3, 0.3, 0.3, 1),
            halign='center',
            valign='middle'
        )
        info_label.bind(size=info_label.setter('text_size'))
        root.add_widget(info_label)

        # Tombol Aksi
        btn = Button(
            text="Klik Saya",
            size_hint=(None, None),
            size=(200, 50),
            pos_hint={'center_x': 0.5},
            background_color=(0.1, 0.6, 0.8, 1)
        )
        btn.bind(on_press=self.on_button_click)
        root.add_widget(btn)

        return root

    def _update_rect(self, instance, value):
        self.rect.pos = instance.pos
        self.rect.size = instance.size

    def on_button_click(self, instance):
        instance.text = "Berhasil Diklik!"

if __name__ == '__main__':
    AppKuApp().run()
```
